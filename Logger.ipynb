```python
#Logging in python
import logging
logging.basicConfig(
    Filename='app.log',
    filemode='r',
    level=logging.DEBUG,
    format='%(asctime)s %(levelname)s %(message)s',
    datefmt='%y-%m-%d'

                    )
logging.debug('This is a debug message')
logging.warning('This is warning message')
logging.critical('This is critical message')
logging.info('This is info message')
logging.error('This is error message')
```

```python
# Creating Multiple Logging
import logging
logging.basicConfig(
    Filename='app.log',
    filemode='r',
    level=logging.DEBUG,
    format='%(asctime)s %(levelname)s %(message)s',
    datefmt='%y-%m-%d'
)
logger1=logging.getLogger('module1')
logger1.setLevel(logging.DEBUG)
logger2=logging.getLogger('module2')
logger2.setLevel(logging.WARNING)
logger1.debug("this is the debug msg")
logger2.warning("This is the warning msg")

```

```python
import logging
logging.basicConfig(
    Filename='app.log',
    filemode='r',
    level=logging.DEBUG,
    format='%(asctime)s %(levelname)s %(message)s',
    datefmt='%y-%m-%d'
)
logger=logging.getLogger('Arithmetic app')
def add(a,b):
  result=a+b
  logger.debug(f'{a}+{b}={result}')
  return result
add(2,3)
```
