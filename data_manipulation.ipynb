```python
import pandas as pd
df=pd.read_csv('sales1_data.csv')
df.head()
```

```python
df.isnull().sum()
```

```python
#handling Missing Values With Mean Imputation
df['Sales'].fillna(df['Sales'].mean())
```

```python
#Renamimhg the Column Names
df.rename(columns={'Sales':'Saless'})
```

```python
#Data  Grouping And Aggregation
df.groupby('Sales') ['Sales'].mean()
```

```python
df.groupby(['Region', 'Sales']).agg({'Sales': 'mean', 'Region': 'count'})
```
