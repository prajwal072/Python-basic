```python
import pandas as pd
#Creating a Series
data=[1,2,3,4,5,6,7,8]
#series=pd.Series(data)
#print(series)
index=['a','b','c','d','e','f','g','h']
series=pd.Series(data,index=index)
print(series)
```

```python
# Creating Series using the Dictionary
dict={
    'name':['prajwal','ashish','ashu'],
    'age':[23,45,66]

}
series=pd.Series(dict)
print(series)
```
