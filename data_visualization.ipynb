```python
import matplotlib.pyplot as plt
#Generating the Line
x=[1,2,3,4,5]
y=[3,4,5,6,7]
plt.plot(x,y,color="green",linestyle="--",marker="o",markersize=10,markerfacecolor="red")
plt.title("Line Graph")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
# Bar Graph
categories=['A','B','C','D']
values=[10,20,15,25]
plt.figure(figsize=(8,5))
plt.bar(categories,values,color='red',edgecolor='black')
plt.title("Bar Graph")
plt.xlabel("Categories")
plt.ylabel("Values")
plt.show()
```

```python
#Scatter Plot
x=[1,2,3,4,5,5,6,7,8]
y=[2,3,4,5,6,7,8,9,10]
plt.figure(figsize=(4,5))
plt.scatter(x,y,color='green',marker='*')
plt.title("Scatter Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")    
plt.show()
```

```python
# Pie chart
labels = ['A', 'B', 'C', 'D']
sizes = [15, 30, 45, 10]    
explode = (0, 0.1, 0, 0)  # explode the second slice
plt.pie(sizes, labels=labels, explode=explode, autopct='%1.1f%%', shadow=True, startangle=140)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
tips=sns.load_dataset('tips')
tips.head()
```

```python
#Scatter Plot
sns.scatterplot(x='total_bill',y='tip',data=tips)
plt.title("Scatter Plot")
plt.show()
```

```python
sns.lineplot(x='day',y='total_bill',data=tips)
plt.title("Line Plot")
plt.show()
```

```python
sns.barplot(x='day',y='total_bill',data=tips)
plt.title("Bar Plot")
plt.show()
```

```python
sns.boxplot(x='day',y='total_bill',data=tips)
plt.title("Box Plot")
plt.show()
```

```python
sns.violinplot(x='day',y='total_bill',data=tips)
plt.title("Violin Plot")
plt.show()
```

```python
#heat map
corr=tips[['total_bill','tip','size']].corr()
sns.heatmap(corr, annot=True)
plt.title("Heat Map")
plt.show()
```
