```python
#Circular Reference
class MyObject:
  def __init__(self,name):
    self.name=name
    print(f"Object {self.name} is created")
  def __Del__(self):
    print(f"Object {self.name} is deleted ")

obj1=MyObject("Object1")
obj2=MyObject("Object2")
obj1.obj2=obj2
obj2.obj1=obj1
del obj1
del obj2
```

```python
#Generators

def generate_numbers(num):
  for i in range(num):

    yield i
for num in generate_numbers(5):
  print(num)
```
