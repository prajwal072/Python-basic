```python
class BankAccount:
  def __init__(self,name,balance):
    self.name=name
    self.balance=balance

  def deposit(self,amount):
    self.balance = self.balance + amount
    print(f"{amount} is credited to account. New balance: {self.balance}")

  def withdraw(self,amount):
    if self.balance >= amount:
      self.balance = self.balance - amount
      print(f"{amount} is debited from account. New balance: {self.balance}")
    else:
      print("Insufficient balance.")

  def checkbalance(self):
    print(f"Current balance for {self.name}: {self.balance}")

bank=BankAccount("prajwal",0)
bank.checkbalance()
bank.deposit(2400)
bank.withdraw(1000)
bank.checkbalance()
bank.withdraw(2000)
```

```python
##Inheritance (Single)
class Car:
  def __init__(self,doors,model,windows):
    self.doors=doors
    self.model=model
    self.windows=windows

  def start(self):
    print(f"{self.model} started running")

class tesla(Car):
  def __init__(self,doors,model,windows):
    super().__init__(doors,model,windows)
    print("The driving is very easy")

Tesla1=tesla(4,"Tesla",4)
Tesla1.start()
```

```python
# Multiple inheritance
class Animal:
  def __init__(self,name):
    self.name=name

  def namy(self):
    print(f"The name of Animal is {self.name}")
class Pet:
  def __init__(self,owner):
    self.owner=owner



  def ownername(self):
    print(f"The name of owner is {self.owner}")

class Dog(Animal,Pet):
  def __init__(self,name,owner):
    Animal.__init__(self,name)
    Pet.__init__(self,owner)

doggy=Dog("luccy","prajwal")
doggy.namy()
doggy.ownername()
```

```python
# Polymorphism
class animal:
  def speak(self):
    print("Animal is Speaking")
class Dog(animal):
  def speak(self):
    print("The Dog is barking")

class Cat(animal):
  def speak(self):
    print("the Cat is Meowing")
# Form that demonstrate the Polymorphism
def animal_speak(animal):
  animal.speak()

dog=Dog()
cat=Cat()
dog.speak()
cat.speak()
animal_speak(dog)
```

```python
# second Example Of Polymorphism
class shape:
  def area(self):
    return"The area of figure"

class rectangle(shape):
  def __init__(self,length,width):
    self.length=length
    self.width=width
  def area(self):
    print(f"The Area is {self.length* self.width}")

class circle(shape):
  def __init__(self,radius):
    self.radius=radius

  def area(self):
    print(f"The area of circle is {3.14*self.radius*self.radius}")

def print_area(shape):
  print(shape.area())

Rectangle=rectangle(4,5)
Circle=circle(5)
print_area(Rectangle)
print_area(Circle)
```

```python
from abc import ABC,abstractmethod
class Vechile(ABC):
  @abstractmethod
  def start_engine(self):
    pass
class Car(Vechile):
  def start_engine(self):
    print("the Car engine started")
class Motor(Vechile):
  def start_engine(self):
    print("The Motor is starting")

def start_engine(vechile):
  vechile.start_engine()

car=Car()
motor=Motor()
start_engine(car)
start_engine(motor)
```

```python
# Abstraction Example
from abc import ABC, abstractmethod

class Vehicle(ABC):
    def drive(self):
        print("the Car is driving")

    @abstractmethod
    def start_engine(self):
        pass

class Car(Vehicle):
    def start_engine(self):
        print("The car engine started")

def operate_car(vehicle):
    vehicle.drive()
    vehicle.start_engine()

car = Car()
operate_car(car)
```

```python
# Magic Methods
class Person:
  def __init__(self,name,age):
    self.name=name
    self.age=age

  def __str__(self):
    return f"{self.name} is {self.age} years old"

person=Person("Prajwal",34)
print(person)
```

```python
#Operator Overloading
class Vector:
  def __init__(self,x,y):
    self.x=x
    self.y=y
  def __add__(self,other):
    return Vector(self.x+other.x,self.y+other.y)
  def __str__(self):
    return f"Vector ({self.x},{self.y})"
v1=Vector(3,4)
v2=Vector(4,5)
print(v1+v2)
```

```python
#Closures Function
def main_welcome():
  msg="Welcome to the main function"
  def sub_main():
    print("Enter in the submain function")
    print(msg)#Accessing the variable from the main function
  return sub_main()
a=main_welcome()
print(a)
```

```python
#Decorator
def main_welcome(func):
  msg="hello All welcome to main function"
  def wrapper(*args, **kwargs):
    print(msg)
    print("hello all welcome to sub main function")
    result = func(*args, **kwargs)
    return result
  return wrapper
```

```python
@main_welcome
def introd():
  print("this is the advanced python course")

introd()
```
